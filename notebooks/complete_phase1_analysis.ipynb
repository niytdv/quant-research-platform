```python
"""
PHASE 1 COMPLETE ANALYSIS
Compare multiple stocks with full metrics and visualizations
"""

from src.data_loader import DataLoader
from src.cleaners import DataCleaner
from src.metrics import FinancialMetrics
from src.visualizer import Visualizer

# Initialize all modules
loader = DataLoader()
cleaner = DataCleaner()
metrics_calc = FinancialMetrics()
viz = Visualizer()

# Define stocks to analyze
tickers = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'SPY']
start_date = '2023-01-01'
end_date = '2024-01-01'

# Load all data
print("Loading data...")
data_dict = loader.fetch_multiple_stocks(tickers, start_date, end_date)

# Clean and align
print("Cleaning data...")
aligned_data = cleaner.align_dataframes(data_dict)

# Calculate returns for all stocks
returns_dict = {}
for ticker, df in aligned_data.items():
    returns_dict[ticker] = metrics_calc.calculate_simple_returns(df)

# CREATE VISUALIZATIONS

# 1. Cumulative Returns Comparison
viz.plot_cumulative_returns(returns_dict)

# 2. Correlation Heatmap
viz.plot_correlation_heatmap(returns_dict)

# 3. Individual Dashboards
for ticker in tickers:
    viz.create_dashboard(aligned_data[ticker], ticker, returns_dict[ticker])

# GENERATE PERFORMANCE REPORT
print("\n" + "="*60)
print("PERFORMANCE REPORT")
print("="*60)

for ticker in tickers:
    print(f"\n{ticker}:")
    print("-" * 40)
    all_metrics = metrics_calc.calculate_all_metrics(aligned_data[ticker])
    
    print(f"Total Return: {all_metrics['total_return']*100:.2f}%")
    print(f"Annualized Return: {all_metrics['annualized_return']*100:.2f}%")
    print(f"Volatility: {all_metrics['annualized_volatility']*100:.2f}%")
    print(f"Sharpe Ratio: {all_metrics['sharpe_ratio']:.2f}")
    print(f"Max Drawdown: {all_metrics['max_drawdown']:.2f}%")

plt.show()
```
