```python
# Week 2: Data Cleaning & Metrics Analysis

from src.data_loader import DataLoader
from src.cleaners import DataCleaner
from src.metrics import FinancialMetrics
import matplotlib.pyplot as plt

# Initialize
loader = DataLoader()
cleaner = DataCleaner()
metrics = FinancialMetrics()

# Load data for multiple stocks
tickers = ['AAPL', 'MSFT', 'GOOGL']
data_dict = loader.fetch_multiple_stocks(tickers, '2023-01-01', '2024-01-01')

# Clean and align
aligned_data = cleaner.align_dataframes(data_dict)

# Calculate metrics for each stock
for ticker, df in aligned_data.items():
    print(f"\n{'='*50}")
    print(f"METRICS FOR {ticker}")
    print(f"{'='*50}")
    
    all_metrics = metrics.calculate_all_metrics(df)
    for key, value in all_metrics.items():
        if isinstance(value, float):
            print(f"{key}: {value:.4f}")
```
